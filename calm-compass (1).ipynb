```python
"""
Agents for Good: Student Mental-Health Support Agent

This notebook defines a simple rule-based mental health support agent for
students and job holders. It is NOT a medical or clinical tool and does not
replace professional help.
"""

from datetime import datetime

class StudentMentalHealthAgent:
    """
    A simple rule-based mental health support agent for students and job holders.
    This is NOT professional medical help. It only gives general emotional support
    and suggestions.
    """

    def __init__(self, user_name="Friend"):
        self.user_name = user_name
        self.conversation_history = []

        # Keywords used to detect crisis-level messages
        self.crisis_keywords = [
            "suicide", "kill myself", "end my life", "no reason to live",
            "self harm", "cut myself", "want to die","life is pointless", "i dont want to live"
        ]

        # Keywords related to stress
        self.stress_keywords = [
            "stress", "pressure", "overwhelmed", "too much work",
            "deadlines", "exams", "tension", "burnout"
        ]

        # Keywords related to anxiety
        self.anxiety_keywords = [
            "anxiety", "panic", "worried", "scared", "restless",
            "overthinking", "nervous", "heart racing"
        ]

        # Keywords related to low mood or depression
        self.depression_keywords = [
            "depressed", "empty", "numb", "sad all the time",
            "no motivation", "tired of everything", "lost interest",
            "worthless", "guilty"
        ]

    def _contains_any(self, text, keywords):
        """
        Check if any keyword appears in the given text.
        """
        text_lower = text.lower()
        return any(keyword in text_lower for keyword in keywords)

    def detect_crisis(self, message):
        """
        Return True if the message looks like a crisis message.
        """
        return self._contains_any(message, self.crisis_keywords)

    def classify_emotion(self, message):
        """
        Rough classification of the main emotional state:
        "stress", "anxiety", "low_mood", or "unclear".
        """
        text_lower = message.lower()

        scores = {
            "stress": sum(1 for kw in self.stress_keywords if kw in text_lower),
            "anxiety": sum(1 for kw in self.anxiety_keywords if kw in text_lower),
            "low_mood": sum(1 for kw in self.depression_keywords if kw in text_lower),
        }

        best_label = max(scores, key=scores.get)
        best_score = scores[best_label]

        if best_score == 0:
            return "unclear", scores

        return best_label, scores

    def supportive_response(self, message):
        """
        Generate a supportive response based on the detected emotion.
        """
        label, scores = self.classify_emotion(message)

        base = (
            "Thank you for sharing this with me, " + self.user_name + ". "
            "Your feelings are valid and it is okay to struggle.\n\n"
        )

        if label == "stress":
            tips = (
                "It sounds like you are experiencing stress.\n"
                "- Try breaking your tasks into small steps.\n"
                "- Take a short break, walk, or stretch away from screens.\n"
                "- If possible, share how you feel with someone you trust.\n"
            )
        elif label == "anxiety":
            tips = (
                "It seems like you are feeling anxious or worried.\n"
                "- Try slow breathing: inhale for 4 seconds, hold for 4, exhale for 6.\n"
                "- Focus only on the next small, manageable task.\n"
                "- Writing your worries down can sometimes reduce the mental load.\n"
            )
        elif label == "low_mood":
            tips = (
                "It seems like you may be feeling low or empty.\n"
                "- Start with one small action, such as listening to music or taking a short walk.\n"
                "- Try to connect with someone you trust, even with a short message.\n"
                "- Do one small activity that you usually enjoy, even if you do not fully feel like it.\n"
            )
        else:
            tips = (
                "I am not fully sure what you are feeling yet.\n"
                "If you want, you can tell me more about what is happening right now.\n"
                "You can also try the sentence: 'I feel ____ because ____.'\n"
            )

        ending = (
            "\nPlease remember that this support agent is not a replacement for a doctor, "
            "therapist, or counselor. If things feel too heavy or out of control, "
            "consider speaking with a mental health professional or a trusted person offline."
        )

        return base + tips + ending

    def crisis_response(self, message):
        """
        Response for possible crisis situations.
        """
        return (
            self.user_name + ", thank you for telling me this.\n\n"
            "What you shared sounds very serious. You deserve immediate support from a real person.\n"
            "Please, as soon as you can, reach out to:\n"
            "- A trusted family member, friend, teacher, or mentor\n"
            "- A mental health professional or counselor\n"
            "- Local emergency services or a crisis helpline if you feel in danger\n\n"
            "You are not alone, and your life is important."
        )

    def respond(self, message):
        """
        Main method: takes a user message and returns the agent's response.
        """
        timestamp = datetime.now().isoformat(timespec="seconds")
        self.conversation_history.append(
            {"time": timestamp, "speaker": "user", "text": message}
        )

        if self.detect_crisis(message):
            reply = self.crisis_response(message)
        else:
            reply = self.supportive_response(message)

        self.conversation_history.append(
            {"time": timestamp, "speaker": "agent", "text": reply}
        )

        return reply

    def show_history(self, n=10):
        """
        Return the last n messages in the conversation history.
        """
        return self.conversation_history[-n:]


# --------------------------------------------------------------------
# Demo section: example usage inside Kaggle Notebook
# --------------------------------------------------------------------

def run_demo():
    """
    Run a simple demo of the StudentMentalHealthAgent.
    """
    agent = StudentMentalHealthAgent(user_name="Charan")

    demo_messages = [
        "I feel a lot of pressure because of my exams and assignments.",
        "I am overthinking about my job and I feel very anxious about my future.",
        "These days I feel empty and have lost interest in things I liked before.",
        "Sometimes I feel like ending my life because nothing seems to work."
    ]

    for i, msg in enumerate(demo_messages, start=1):
        print("Example", i)
        print("USER : ", msg)
        reply = agent.respond(msg)
        print("AGENT: ", reply)
        print("-" * 80)


# Run the demo automatically when this cell is executed
if __name__ == "__main__":
    print("Running demo for Student Mental-Health Support Agent...\n")
    run_demo()
    print("\nDemo finished. You can create your own agent and messages below if you like.")
```

```python
from datetime import datetime

class StudentMentalHealthAgent:
    """
    A rule-based mental health support agent with severity scoring.
    This is NOT professional help and cannot replace real mental health care.
    """

    def __init__(self, user_name="Friend"):
        self.user_name = user_name

        # Crisis-related keywords (very high severity)
        self.crisis_keywords = [
            "suicide", "kill myself", "end my life", "ending my life",
            "no reason to live", "no point in living",
            "self harm", "self-harm", "cut myself",
            "hurt myself", "want to die", "wish i was dead",
            "i dont want to live", "life is pointless", "give up on life"
        ]

        # Emotional keywords
        self.stress_keywords = [
            "stress", "pressure", "overwhelmed", "too much work",
            "deadlines", "exams", "tension", "burnout"
        ]

        self.anxiety_keywords = [
            "anxiety", "panic", "worried", "scared", "restless",
            "overthinking", "nervous", "heart racing"
        ]

        self.depression_keywords = [
            "depressed", "empty", "numb", "sad all the time",
            "no motivation", "tired", "lost interest",
            "worthless", "guilty", "hopeless"
        ]

    # ----------------------------
    # Keyword count helper
    # ----------------------------
    def _count_keywords(self, message, keywords):
        msg = message.lower()
        return sum(1 for kw in keywords if kw in msg)

    # ----------------------------
    # Severity Scoring System
    # ----------------------------
    def calculate_severity(self, message):
        score = 0
        msg = message.lower()

        # Crisis instantly maxes severity
        if any(kw in msg for kw in self.crisis_keywords):
            return 12

        # Weighted scoring
        score += 2 * self._count_keywords(msg, self.stress_keywords)
        score += 3 * self._count_keywords(msg, self.anxiety_keywords)
        score += 4 * self._count_keywords(msg, self.depression_keywords)

        return score

    # ----------------------------
    # Response Generators
    # ----------------------------
    def crisis_response(self):
        return (
            f"{self.user_name}, thank you for telling me this.\n\n"
            "What you shared sounds very serious. You deserve immediate support.\n"
            "Please reach out right now to:\n"
            "- A trusted friend or family member\n"
            "- A counselor or mental health professional\n"
            "- Emergency services or a crisis helpline\n\n"
            "You are not alone, and your life matters."
        )

    def mild_response(self):
        return (
            "It sounds like something is bothering you. "
            "Your feelings are important, and I am here to listen."
        )

    def moderate_response(self):
        return (
            "I can sense you are going through something difficult. "
            "Try taking a moment to breathe and understand how you feel.\n"
            "You are not alone."
        )

    def high_response(self):
        return (
            "It seems like you are dealing with strong emotional pressure. "
            "This must be very hard for you.\n"
            "Talking to someone you trust may help you feel supported."
        )

    # ----------------------------
    # Main Response Logic
    # ----------------------------
    def respond(self, message):
        severity = self.calculate_severity(message)

        # Crisis case
        if severity >= 10:
            return self.crisis_response()

        # High severity
        elif severity >= 6:
            return (
                f"Thank you for sharing this with me, {self.user_name}.\n\n"
                + self.high_response()
                + "\n\nRemember, you matter and support is available."
            )

        # Moderate severity
        elif severity >= 3:
            return (
                f"Thanks for reaching out, {self.user_name}.\n\n"
                + self.moderate_response()
                + "\n\nI am here with you."
            )

        # Mild severity
        else:
            return (
                f"Thank you for telling me this, {self.user_name}.\n\n"
                + self.mild_response()
            )

# -------------------------------------
# DEMO
# -------------------------------------

def run_demo():
    agent = StudentMentalHealthAgent("Charan")
    tests = [
        "I feel stressed with exams.",
        "I am anxious and overthinking everything.",
        "I feel empty and hopeless.",
        "Sometimes I feel like ending my life."
    ]
    for i, t in enumerate(tests, 1):
        print(f"\nExample {i}")
        print("USER:", t)
        print("AGENT:", agent.respond(t))
        print("-" * 80)

# Run demo
run_demo()
```
